## Project Overview
This notebook presents the complete machine learning pipeline for detecting spam emails using Logistic Regression trained on the Enron Spam Dataset (33,716 emails), achieving 99.00% accuracy.

### Pipeline Steps:
1. Data Loading and Exploration
2. Text Preprocessing
3. TF-IDF Feature Extraction
4. Model Training (4 algorithms compared)
5. Evaluation and Results
6. Model Export

## Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re, json, time, warnings
warnings.filterwarnings('ignore')
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix
import joblib
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
print('All libraries imported!')
```

## Step 2: Load Dataset
The Enron Spam Dataset contains 33,716 real labeled emails (spam and legitimate).

```python
df = pd.read_csv('../enron_spam_data.csv')
print(f'Total emails: {len(df):,}')
print(f'Columns: {list(df.columns)}')
print(f'\nClass distribution:')
print(df['Spam/Ham'].value_counts())
df.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
dist = df['Spam/Ham'].value_counts()
dist.plot(kind='bar', ax=axes[0], color=['#e74c3c','#2ecc71'], edgecolor='black')
axes[0].set_title('Email Class Distribution', fontweight='bold')
axes[1].pie(dist, labels=dist.index, colors=['#e74c3c','#2ecc71'], autopct='%1.1f%%')
axes[1].set_title('Spam vs Ham Ratio', fontweight='bold')
plt.suptitle('Enron Dataset Analysis', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Step 3: Text Preprocessing
Pipeline: Lowercase -> Remove HTML/URLs -> Remove special chars -> Stopword removal -> Lemmatization

```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    if not isinstance(text, str): return ''
    text = text.lower()
    text = re.sub(r'<[^>]+>', ' ', text)
    text = re.sub(r'http\S+|www\S+', ' ', text)
    text = re.sub(r'\S+@\S+', ' ', text)
    text = re.sub(r'[^a-z\s]', ' ', text)
    words = [lemmatizer.lemmatize(w) for w in text.split() if w not in stop_words and len(w) > 2]
    return ' '.join(words)

sample = "CONGRATULATIONS! You've WON $1,000,000!!! Click here: http://win.com NOW!!!"
print('Original:', sample)
print('Processed:', preprocess_text(sample))

print('\nPreprocessing all emails...')
df['processed_text'] = df['Message'].apply(preprocess_text)
df['label'] = (df['Spam/Ham'] == 'spam').astype(int)
print('Done!')
```

## Step 4: TF-IDF Feature Extraction
Converts email text into 10,000 numerical features using Term Frequency-Inverse Document Frequency.

```python
X_train, X_test, y_train, y_test = train_test_split(df['processed_text'], df['label'], test_size=0.2, random_state=42, stratify=df['label'])
vectorizer = TfidfVectorizer(max_features=10000, ngram_range=(1, 2), min_df=2)
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf  = vectorizer.transform(X_test)
print(f'Training samples: {len(X_train):,}')
print(f'Testing samples:  {len(X_test):,}')
print(f'Feature matrix shape: {X_train_tfidf.shape}')
```

## Step 5: Model Training and Comparison
Four supervised learning algorithms are trained and evaluated.

```python
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Support Vector Machine': LinearSVC(max_iter=2000, random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    'Multinomial Naive Bayes': MultinomialNB(alpha=0.1)
}
results = {}
print(f'{"Model":<30} {"Accuracy":>10} {"Precision":>10} {"Recall":>10} {"F1":>10}')
print('-' * 65)
for name, model in models.items():
    model.fit(X_train_tfidf, y_train)
    y_pred = model.predict(X_test_tfidf)
    results[name] = {'model': model, 'predictions': y_pred,
        'accuracy': accuracy_score(y_test, y_pred),
        'precision': precision_score(y_test, y_pred),
        'recall': recall_score(y_test, y_pred),
        'f1': f1_score(y_test, y_pred)}
    r = results[name]
    print(f'{name:<30} {r["accuracy"]:>10.4f} {r["precision"]:>10.4f} {r["recall"]:>10.4f} {r["f1"]:>10.4f}')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
for idx, (name, r) in enumerate(results.items()):
    ax = axes[idx // 2][idx % 2]
    cm = confusion_matrix(y_test, r['predictions'])
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax, xticklabels=['Ham','Spam'], yticklabels=['Ham','Spam'])
    ax.set_title(f'{name}\nAccuracy: {r["accuracy"]*100:.2f}%', fontweight='bold')
    ax.set_xlabel('Predicted'); ax.set_ylabel('Actual')
plt.suptitle('Confusion Matrices - All Models', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
best = results['Logistic Regression']
print('=== Best Model: Logistic Regression ===')
print(classification_report(y_test, best['predictions'], target_names=['Ham', 'Spam']))
print(f'Final Accuracy: {best["accuracy"]*100:.2f}%')
```

## Step 6: Export Model
Save the trained model and vectorizer for use in the live web application.

```python
joblib.dump(results['Logistic Regression']['model'], 'logistic_regression_final.joblib')
joblib.dump(vectorizer, 'tfidf_vectorizer_final.joblib')
print('Model saved: logistic_regression_final.joblib')
print('Vectorizer saved: tfidf_vectorizer_final.joblib')
print('\nLive demo: https://bouestispamdetector.com.ng')
```
